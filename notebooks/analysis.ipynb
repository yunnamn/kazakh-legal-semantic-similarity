# Domain Adaptation for Kazakh Legal Semantic Similarity\n\nThis lightweight notebook imports project modules for inspection, plotting, and qualitative review. Core implementation lives in `src/` and CLI scripts.

```python
from pathlib import Path
import sys
import pandas as pd

def find_repo_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / 'src').is_dir() and (candidate / 'config' / 'default.yaml').exists():
            return candidate
    raise RuntimeError('Could not locate repository root.')

repo_root = find_repo_root(Path.cwd().resolve())
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from src.utils import load_config, get_path
from src.corpus import load_corpus, load_split, corpus_statistics
from src.evaluation import plot_sts_metrics, plot_score_scatter

config = load_config(repo_root / 'config' / 'default.yaml')
repo_root
```

## Corpus Inspection

```python
try:
    corpus = load_corpus(config)
    display(corpus.head())
    corpus_statistics(corpus, config)
except FileNotFoundError as exc:
    print(exc)
```

## Split Counts

```python
for split in ['train', 'validation', 'mlm_test', 'annotation_pool']:
    try:
        df = load_split(split, config)
        print(split, len(df))
    except FileNotFoundError:
        print(split, 'not built yet')
```

## Measured Results

```python
metrics_path = get_path(config, 'metrics_dir') / 'sts_metrics.csv'
if metrics_path.exists():
    metrics = pd.read_csv(metrics_path)
    display(metrics)
    plot_sts_metrics(metrics, config)
else:
    print('Semantic-similarity evaluation is pending manual annotation.')
```
